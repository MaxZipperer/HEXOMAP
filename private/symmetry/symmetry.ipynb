```python
# These are the Crystal Frame operators as matrices
def GetSymRotMat_new(symtype='Cubic'):
    if symtype in [str(spg) for spg in range(1,2+1)]:
        # Triclinic
        m = np.zeros((1,3,3))
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        return m
    
    elif symtype in [str(spg) for spg in range(3,15+1)]:
        # Monoclinic
        m = np.zeros((2,3,3))
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || b
        m[1] = [[-1,0,0],
                [0,1,0],
                [0,0,-1]]
        return m
    
    elif symtype in [str(spg) for spg in range(16,74+1)]:
        # Orthorhombic
        m = np.zeros((4,3,3))
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || a
        m[1] = [[1,0,0],
                [0,-1,0],
                [0,0,-1]]    
        # 2-fold || b
        m[2] = [[-1,0,0],
                [0,1,0],
                [0,0,-1]]
        # 2-fold || c
        m[3] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]
        return m
        
    elif symtype in [str(spg) for spg in range(75,88+1)]:
        m = np.zeros((4,3,3))
        # Tetragonal_lower
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || c
        m[1] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]    
        # 4-fold || c
        m[2] = [[0,-1,0],
                [1,0,0],
                [0,0,1]]
        # 4-fold || c
        m[3] = m[2].T
        return m
        
    elif symtype == 'Tetragonal' or symtype in [str(spg) for spg in range(89,142+1)]:
        m = np.zeros((8, 3, 3))
        # Tetragonal
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || c
        m[1] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]    
        # 4-fold || c
        m[2] = [[0,-1,0],
                [1,0,0],
                [0,0,1]]
        # 4-fold || c
        m[3] = m[2].T
        # 2-fold || a
        m[4] = [[1,0,0],
                [0,-1,0],
                [0,0,-1]]    
        # 2-fold || b
        m[5] = [[-1,0,0],
                [0,1,0],
                [0,0,-1]]
        # 2-fold || [110]
        m[6] = [[0,1,0],
                [1,0,0],
                [0,0,-1]]
        # 2-fold || [1-10]
        m[7] = [[0,-1,0],
                [-1,0,0],
                [0,0,-1]]
        return m

    elif symtype in [str(spg) for spg in range(143,148+1)]:
        m = np.zeros((3,3,3))
        # Trigonal_lower
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 3-fold || [111]
        m[1] = [[0,0,1],
                [1,0,0],
                [0,1,0]]
        # 3-fold || [111]
        m[2] = m[1].T
        return m

    elif symtype in [str(spg) for spg in range(149,167+1)]:
        m = np.zeros((6,3,3))
        # Trigonal
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 3-fold || [111]
        m[1] = [[0,0,1],
                [1,0,0],
                [0,1,0]]
        # 3-fold || [111]
        m[2] = m[1].T
        # 2-fold || a
        m[3] = [[1,0,0],
                [0,-1,0],
                [0,0,-1]]    
        # 2-fold || b
        m[4] = [[-1,0,0],
                [0,1,0],
                [0,0,-1]]
        # 2-fold || c
        m[5] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]
        return m

    elif symtype in [str(spg) for spg in range(168,176+1)]:
        m = np.zeros((6,3,3))
        # Hexagonal_lower
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || c
        m[1] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]
        # 3-fold || c
        m[2] = [[0,-1,0],
                [1,-1,0],
                [0,0,1]]
        # 3-fold || c
        m[3] = [[-1,1,0],
                [-1,0,0],
                [0,0,1]]
        # 6-fold || c
        m[4] = [[1,-1,0],
                [1,0,0],
                [0,0,1]]
        # 6-fold || c
        m[5] = [[0,1,0],
                [-1,1,0],
                [0,0,1]]
        return m

    elif symtype == 'Hexagonal' or symtype in [str(spg) for spg in range(177,194+1)]:
        m = np.zeros((12, 3, 3))
        # Hexagonal
        # Identity
        m[0] = [[1,0,0],
                [0,1,0],
                [0,0,1]]
        # 2-fold || c
        m[1] = [[-1,0,0],
                [0,-1,0],
                [0,0,1]]
        # 3-fold || c
        m[2] = [[0,-1,0],
                [1,-1,0],
                [0,0,1]]
        # 3-fold || c
        m[3] = [[-1,1,0],
                [-1,0,0],
                [0,0,1]]
        # 6-fold || c
        m[4] = [[1,-1,0],
                [1,0,0],
                [0,0,1]]
        # 6-fold || c
        m[5] = [[0,1,0],
                [-1,1,0],
                [0,0,1]]
        # 2-fold || a
        m[6] = [[1,-1,0],
                [0,-1,0],
                [0,0,-1]]
        # 2-fold || b
        m[7] = [[-1,0,0],
                [-1,1,0],
                [0,0,-1]]
        # 2-fold || [110]
        m[8] = [[0,1,0],
                [1,0,0],
                [0,0,-1]]
        # 2-fold || [1-10]
        m[9] = [[0,-1,0],
                [-1,0,0],
                [0,0,-1]]
        # 2-fold || [210]
        m[10] = [[1,0,0],
                [1,-1,0],
                [0,0,-1]]
        # 2-fold || [120]
        m[11] = [[-1,1,0],
                [0,1,0],
                [0,0,-1]]
        return m
        
    elif symtype in [str(spg) for spg in range(195,206+1)]:
        m = np.zeros((12,3,3))
        # Cubic_lower
        # Identity
        m[0] = [[1,0,0],
               [0,1,0],
               [0,0,1]]
        # 2-fold || a
        m[1] = [[1,0,0],
               [0,-1,0],
               [0,0,-1]]
        # 2-fold || b
        m[2] = [[-1,0,0],
               [0,1,0],
               [0,0,-1]]
        # 2-fold || c
        m[3] = [[-1,0,0],
               [0,-1,0],
               [0,0,1]]
        # 3-fold || [111]
        m[4] = [[0,0,1],
               [1,0,0],
               [0,1,0]]
        # 3-fold || [111]
        m[5] = m[4].T
        # 3-fold || [1-1-1]
        m[6] = [[0,0,-1],
               [-1,0,0],
               [0,1,0]]
        # 3-fold || [1-1-1]
        m[7] = m[6].T
        # 3-fold || [-11-1]
        m[8] = [[0,0,1],
               [-1,0,0],
               [0,-1,0]]
        # 3-fold || [-11-1]
        m[9] = m[8].T
        # 3-fold || [-1-11]
        m[10] = [[0,0,-1],
               [1,0,0],
               [0,-1,0]]
        # 3-fold || [-1-11]
        m[11] = m[10].T
        return m

    elif symtype == 'Cubic' or symtype in [str(spg) for spg in range(207,230+1)]:
        m = np.zeros((24, 3, 3))
        # Cubic
        # Identity
        m[0] = [[1,0,0],
               [0,1,0],
               [0,0,1]]
        # 2-fold || a
        m[1] = [[1,0,0],
               [0,-1,0],
               [0,0,-1]]
        # 2-fold || b
        m[2] = [[-1,0,0],
               [0,1,0],
               [0,0,-1]]
        # 2-fold || c
        m[3] = [[-1,0,0],
               [0,-1,0],
               [0,0,1]]
        # 3-fold || [111]
        m[4] = [[0,0,1],
               [1,0,0],
               [0,1,0]]
        # 3-fold || [111]
        m[5] = m[4].T
        # 3-fold || [1-1-1]
        m[6] = [[0,0,-1],
               [-1,0,0],
               [0,1,0]]
        # 3-fold || [1-1-1]
        m[7] = m[6].T
        # 3-fold || [-11-1]
        m[8] = [[0,0,1],
               [-1,0,0],
               [0,-1,0]]
        # 3-fold || [-11-1]
        m[9] = m[8].T
        # 3-fold || [-1-11]
        m[10] = [[0,0,-1],
               [1,0,0],
               [0,-1,0]]
        # 3-fold || [-1-11]
        m[11] = m[10].T
        # 2-fold || [110]
        m[12] = [[0,1,0],
                 [1,0,0],
                 [0,0,-1]]
        # 2-fold || [1-10]
        m[13] = [[0,-1,0],
                 [-1,0,0],
                 [0,0,-1]]
        # 2-fold || [101]
        m[14] = [[0,0,1],
                 [0,-1,0],
                 [1,0,0]]
        # 2-fold || [-101]
        m[15] = [[0,0,-1],
                 [0,-1,0],
                 [-1,0,0]]
        # 2-fold || [011]
        m[16] = [[-1,0,0],
                 [0,0,1],
                 [0,1,0]]
        # 2-fold || [01-1]
        m[17] = [[-1,0,0],
                 [0,0,-1],
                 [0,-1,0]]
        # 4-fold || a
        m[18] = [[1,0,0],
                 [0,0,-1],
                 [0,1,0]]
        # 4-fold || a
        m[19] = m[18].T
        # 4-fold || b
        m[20] = [[0,0,1],
                 [0,1,0],
                 [-1,0,0]]
        # 4-fold || b
        m[21] = m[20].T
        # 4-fold || c
        m[22] = [[0,-1,0],
                 [1,0,0],
                 [0,0,1]]
        # 4-fold || c
        m[23] = m[22].T
        return m
```

```python
def directStructureMatrix(a_vec, b_vec, c_vec):
    dsm = np.column_stack([a_vec, b_vec, c_vec])
    return dsm
def op2Cart(op, dsm):
    op_cart = np.dot(dsm, np.dot(op, np.linalg.inv(dsm)))
    return op_cart
```

```python
a_vec, b_vec, c_vec = getLatticeVectors(2.95,2.95,4.68,90,90,120)
dsm = directStructureMatrix(a_vec, b_vec, c_vec)
op_crys = GetSymRotMat_new('194')
num = op_crys.shape[0]
op_cart = np.zeros((num,3,3))
for op in range(num):
    op_cart[op] = op2Cart(op_crys[op], dsm)
print(np.round(op_cart,5))
```
